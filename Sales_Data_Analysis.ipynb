```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Product": [
        "Laptop", "Mobile", "Headphones",
        "Laptop", "Mobile", "Keyboard",
        "Headphones", "Laptop", "Keyboard", "Mobile"
    ],
    "Category": [
        "Electronics", "Electronics", "Accessories",
        "Electronics", "Electronics", "Accessories",
        "Accessories", "Electronics", "Accessories", "Electronics"
    ],
    "Quantity": [2, 5, 10, 1, 3, 6, 4, 2, 8, 4],
    "Price": [50000, 15000, 1000, 50000, 15000, 800, 1000, 50000, 800, 15000]
}

df = pd.DataFrame(data)

df["Sales"] = df["Quantity"] * df["Price"]

df.to_csv("sales_data.csv", index=False)

print("CSV file created successfully!")
print(df)
```

```python
print("Dataset Information:")
print(df.info())

print("\nTotal Rows and Columns:")
print(df.shape)

print("\nTotal Sales:")
print(df["Sales"].sum())

print("\nAverage Sales:")
print(df["Sales"].mean())

print("\nProduct-wise Sales:")
print(df.groupby("Product")["Sales"].sum())
```

```python
product_sales = df.groupby("Product")["Sales"].sum()

product_sales.plot(
    kind="bar",
    title="Product-wise Sales Analysis",
    xlabel="Product",
    ylabel="Total Sales",
    rot=45
)

plt.tight_layout()
plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

category_sales.plot(
    kind="pie",
    autopct="%1.1f%%",
    title="Category-wise Sales Distribution"
)

plt.ylabel("")
plt.show()
```

```python
print("===== FINAL SALES REPORT =====")

print("Total Sales:", df["Sales"].sum())

print(
    "Best Selling Product:",
    df.groupby("Product")["Quantity"].sum().idxmax()
)

print(
    "Highest Revenue Product:",
    df.groupby("Product")["Sales"].sum().idxmax()
)

print("Analysis Completed Successfully!")
```
